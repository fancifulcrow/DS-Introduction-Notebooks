## Importing Matplotlib

Inside Matplotlib, there is a module called `pyplot`, which is what we mostly use. It is commonly imported with the alias `plt`.

````python
import pandas as pd
import matplotlib.pyplot as plt
````

## Types of Charts

### Line Chart

A line chart is used to show trends or continuous data. It is especially useful when tracking changes over time. You create a line plot using the `plot()` function.

````python
df = pd.DataFrame({
    "Month": ["Jan", "Feb", "Mar", "Apr", "May", "Jun", "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"],
    "Revenue": [12, 25, 13, 13, 15, 20, 30, 27, 21, 10, 5, 2],
    "Expenses": [6, 10, 7, 8, 10, 9, 8, 15, 21, 10, 3, 5]
})
````

````python
plt.plot(df["Month"], df["Revenue"])
plt.show()
````

We are capable of customizing many aspects of a graph in Matplotlib, which makes it very powerful for creating clear and visually meaningful visualizations. Beyond just plotting data, you can adjust colors, add markers, change line styles, control labels, modify titles, and fine-tune almost every visual element to better communicate what your data is showing.

````python
plt.figure(figsize=(8, 4))
plt.plot(df["Month"], df["Revenue"], color="red", marker="o", linestyle="--")
plt.title("Revenue Throughout the Year", fontsize=12, fontweight="bold")
plt.xlabel("Month")
plt.ylabel("Revenue")
plt.xticks(rotation=45)
plt.grid(True)
plt.show()
````

When plotting multiple lines or dots on the same graph, use the `label` parameter to assign a name to each line, and call `plt.legend()` to display them. Without `plt.legend()`, the labels will not appear on the graph.

````python
plt.plot(df["Month"], df["Revenue"], color="blue", label="Revenue")
plt.plot(df["Month"], df["Expenses"], color="red", label="Expenses")
plt.title("Revenue and Expenses Over Time", fontsize=12, fontweight="bold")
plt.xlabel("Month")
plt.ylabel("Amount")
plt.xticks(rotation=45)
plt.legend()
plt.show()
````

### Scatter Plot

A scatter plot is used to show the relationship between two numerical variables. Unlike a line plot, it does not connect points with lines. Instead, it displays individual points, which makes it useful for spotting clusters, correlations, or outliers. The function used is `scatter()`.

````python
df = pd.DataFrame({
    "Height": [150, 152, 155, 158, 160, 162, 165, 167, 168, 170,
               172, 174, 175, 176, 178, 180, 182, 184, 186, 188],
    "Weight": [50, 52, 54, 55, 60, 58, 62, 64, 66, 70,
               68, 71, 72, 74, 76, 78, 80, 82, 84, 86]
})
````

````python
plt.scatter(df["Height"], df["Weight"])
plt.show()
````

Similarly to the line chart, you can customize the plot.

````python
plt.scatter(df["Height"], df["Weight"], color="#32B140", marker="x", s=100)
plt.title("Height vs. Weight", fontsize=12, fontweight="bold")
plt.xlabel("Height")
plt.ylabel("Weight")
plt.grid(True, alpha=0.5)
plt.show()
````

### Bar Chart

A bar chart visually represents data using rectangular bars or columns. The length of each bar corresponds proportionally to its value. You can present these bars either horizontally or vertically.

You create bar charts using the `bar()` function, or horizontal bar charts using `barh()`.

````python
df = pd.DataFrame({
    "Fruit": ["Apple", "Banana", "Cherry", "Dragon fruit"],
    "Sales": [30, 20, 25, 35]
})
````

````python
plt.bar(df["Fruit"], df["Sales"])
plt.show()
````

````python
plt.barh(df["Fruit"], df["Sales"])
plt.show()
````

You can style bar charts the same way as other chart types.

````python
plt.bar(df["Fruit"], df["Sales"], color="coral", edgecolor="black", linewidth=1.2, alpha=0.9)
plt.title("Fruit Sales", fontsize=12, fontweight="bold")
plt.xlabel("Fruit")
plt.ylabel("Sales")
plt.grid(True, axis="y", alpha=0.3)
plt.show()
````

````python
plt.barh(df["Fruit"], df["Sales"], color="#84E4EB", edgecolor="black", linewidth=1.2)
plt.title("Fruit Sales", fontsize=12, fontweight="bold")
plt.xlabel("Sales")
plt.ylabel("Fruit")
plt.show()
````

### Pie Chart

A pie chart is used to show proportions or percentages of a whole. Each slice of the pie represents a category's contribution to the total. The function used is `pie()`, and it is most effective when the data is limited to a small number of categories.

The `pie()` function expects a sequence of values, so it is common to first aggregate your data (e.g. using `value_counts()`) before plotting.

````python
df = pd.DataFrame({
    "Fruit": ["Apple", "Banana", "Cherry", "Apple", "Apple", "Banana", "Apple", "Banana"]
})

counts = df["Fruit"].value_counts()

print(counts)
````

````python
plt.pie(x=counts.values, labels=counts.index)
plt.show()
````

This chart can also be customized.

````python
plt.figure(figsize=(6, 6))
plt.pie(x=counts.values, labels=counts.index, colors=["#4C72B0", "#55A868", "#C44E52"], autopct="%.1f%%")
plt.title("Fruits", fontsize=16, fontweight="bold")
plt.legend()
plt.show()
````

## Using Pandas Built-in Plotting Functions

Pandas integrates Matplotlib under the hood. You can plot directly from a DataFrame or Series using built-in methods like `df.plot()`. This makes data visualization faster, especially when working with structured datasets. You just need to specify the `kind` of the plot.

````python
df = pd.DataFrame({
    "City": ["Lagos", "New York", "London", "Tokyo", "Paris", "Prague"],
    "Population (in millions)": [17.8, 8.5, 9.1, 14.3, 2.1, 1.4]
})
````

````python
df.plot(x="City", y="Population (in millions)", kind="bar", width=0.8, color="#9a53d0")
plt.title("Population of Cities", fontsize=12, fontweight="bold")
plt.xlabel("City")
plt.ylabel("Population (in millions)")
plt.xticks(rotation=45)
plt.show()
````
