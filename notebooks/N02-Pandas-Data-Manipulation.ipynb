## Selection

### Selecting Columns

You can select a column with:

````python
print(df.name)
````

However, if you want to access a column with a name that has spaces or special characters, you can use the bracket notation. This way is usually preferred.

````python
# Single column
print(df["name"])
````

````python
# Multiple Columns
print(df[["name", "salary", "department"]])
````

When you write `df["name"]`, you get back a Series. Think of it as pulling out a single column from the table. When you write `df[["name", "salary", "department"]]` (double brackets), you get back a DataFrame, which is just a mini-table with only those columns.

### Selecting Rows: `.loc` and `.iloc`

The difference is that `.loc` thinks in labels while `.iloc` thinks in positions. 

`.loc` uses the actual index label and column name, so `df.loc[2, "salary"]` means "row labelled 2, column named salary", and slicing with `df.loc[1:3]` is inclusive on both ends, giving you rows 1, 2, and 3. 

`.iloc` on the other hand uses integer positions like a list index, so `df.iloc[0, 2]` means "first row, third column" regardless of what the column is named, and `df.iloc[1:3]` follows normal Python slicing rules where the right end is exclusive, giving you only positions 1 and 2.

| Method | Uses | Example |
|--------|------|---------|
| `.loc` | **Label-based:** index labels and column names | `df.loc[0, "name"]` |
| `.iloc` | **Integer-based:** row/column position numbers | `df.iloc[0, 1]` |

````python
print(df.loc[2])
````

````python
print(df.loc[0, "name"])
````

````python
print(df.loc[1:3, ["name", "salary"]])
````

````python
print(df.iloc[0, 1])
````

````python
print(df.iloc[0:3, 0:3])
````

````python
print(df.iloc[-1])
````

## Converting Datatypes

````python
print(df.dtypes)
````

### `astype()`

We can convert between datatypes using `astype()`. Take the `salary` column for example. Currently it is stored as `int6`4, but we can convert it to float.

````python
df["salary"] = df["salary"].astype(float)

print(df["salary"].dtype)
````

We will also convert the `department` column to the `category` type. This makes sense for columns that have a small number of repeating values (like "Engineering", "Marketing", "HR"). Storing them as category instead of plain strings uses less memory and can also make certain operations faster.

````python
df["department"] = df["department"].astype("category")

print(df["department"].dtype)
````

### Converting from Strings to Numbers

When you read data from a CSV or Excel file, pandas sometimes loads number columns as strings, especially if the column had mixed content like `"42"`, `"N/A"`, or `"$100"`. Calling `astype(float)` on a column like that will crash immediately if it hits any value it cannot convert.

`pd.to_numeric()` is the safer alternative because it is specifically designed for converting to numbers and gives you control over what happens when it encounters a bad value through the errors parameter, which has three options:

- `errors="raise"`:the default, crashes on bad values just like `astype()`
- `errors="coerce"`: converts bad values to `NaN` instead of crashing, and carries on
- `errors="ignore"`: leaves bad values as they are and carries on

You may notice that our `rating` column is stored as a string. Since we want to perform numerical operations on it, let us convert it to a number.

````python
print(df["rating"].dtype)
````

````python
df["rating"] = pd.to_numeric(df["rating"], errors="coerce")

print(df["rating"].dtype)
````

### Working with Datetime

Datetime is one of the most important types in data analysis because dates appear in almost every real dataset: order dates, sign-up times, transaction records and so on. The problem is that sometimes, when pandas loads a file, it may read date columns as plain strings, so you cannot do things like extract the month or calculate the gap between two dates. Fortunately, we can convert to datetime using `pd.to_datetime()`.

````python
print(df["join_date"].dtype)
````

````python
df["join_date"] = pd.to_datetime(df["join_date"], errors="coerce")

print(df["join_date"].dtype)
print(df["join_date"])
````

````python
# The .dt accessor gives you datetime components
print(df["join_date"].dt.year)
````

````python
print(df["join_date"].dt.day_name())
````

### Working with Strings

Pandas also gives you a `.str` accessor for running string operations across an entire column at once, no loop needed. It works a lot like Python's normal string methods, just applied row by row.

````python
# Lowercase every value in the column
print(df["name"].str.lower())
````

## Data Transformation

### Creating New Columns

Creating a new column is as easy as assigning to a column name that doesn't exist yet. Pandas will apply the operation to every row for you, no loop needed.

````python
df["bonus"] = df["salary"] * 0.1

print(df[["name", "salary", "bonus"]])
````

### Mathematical Operations

You can do math directly between columns too, since pandas lines up the rows automatically. This is called a vectorized operation, and it's much faster than looping through rows one by one.

````python
df["total_comp"] = df["salary"] + df["bonus"]

print(df[["name", "salary", "bonus", "total_comp"]])
````

The usual operators all work: `+`, `-`, `*`, `/`, `**` (power). You can also compare columns, which gives you back a Series of `True`/`False` values.

````python
print(df["years"] > 5)
````

### `map()` and `apply()`

`map()` works on a single Series and is great for simple, value-by-value substitution, usually with a dictionary or a small function.

`apply()` is more general. It works on a Series or a whole DataFrame, and can run any function you write, even ones that look at multiple columns at once (using `axis=1` to apply row by row).

````python
dept_codes = {"Engineering": "ENG", "Marketing": "MKT", "HR": "HR"}

df["dept_code"] = df["department"].map(dept_codes)

print(df[["name", "department", "dept_code"]])
````

````python
def performance_label(row):
    if row["rating"] >= 4.3:
        return "Top performer"
    elif row["rating"] >= 4.0:
        return "Strong"
    else:
        return "Needs review"

df["performance"] = df.apply(performance_label, axis=1)

print(df[["name", "rating", "performance"]])
````

## Grouping and Aggregation

### Aggregate Functions

You can call functions like `mean()`, `sum()`, `count()`, `min()` and `max()` directly on colums.

````python
# Average salary
print(df["salary"].mean())
````

````python
# Lowest rating
print(df["rating"].min())
````

### `value_counts`

`value_counts()` is a quick way to see how many times each unique value shows up in a column. It is usually the first thing to run on a categorical column.

````python
print(df["department"].value_counts())
````

### `groupby()`

`groupby()` splits the DataFrame into groups based on the values in a column. On its own, it does not return anything useful to look at. You need to chain an aggregation onto it to actually combine the groups back into a result.

````python
# Average salary by department
print(df.groupby("department")["salary"].mean())
````

````python
# Count of employees by department
print(df.groupby("department")["name"].count())
````

## Conditional Filtering

Filtering means keeping only the rows that match some condition, for example only employees earning above a certain salary. The idea is always the same: write a condition that produces a `True`/`False` value for every row, then use that inside the square brackets to keep only the `True` rows.

### Boolean Indexing

`df["salary"] > 80000` on its own gives you a Series of `True`/`False` values, one per row. Placing that inside `df[...]` filters the DataFrame down to only the rows where the condition was `True`.

````python
high_earners = df[df["salary"] > 80000]

print(high_earners[["name", "department", "salary"]])
````

### Combining Conditions with `&` and `|`

To combine more than one condition, use `&` for "and" and `|` for "or". You cannot use the plain Python `and`/`or` keywords here, they do not work on a whole Series. Each condition also needs its own set of parentheses, otherwise Python tries to evaluate the operators in the wrong order.

````python
# Filter employees in the Engineering department with a salary greater than 80000
result = df[(df["salary"] > 80000) & (df["department"] == "Engineering")]

print(result[["name", "department", "salary"]])
````

````python
# Filter employees in the HR or Marketing departments
result = df[(df["department"] == "HR") | (df["department"] == "Marketing")]

print(result[["name", "department", "salary"]])
````

### `query()`

`query()` lets you write the same kind of filter as a plain string, which some people find easier to read. Column names are used directly with no need for `df["..."]`, and you can refer to an outside Python variable by putting an `@` in front of it.

````python
result = df.query("salary > 80000 and department == 'Engineering'")

print(result[["name", "department", "salary"]])
````

````python
min_salary = 80000

result = df.query("salary > @min_salary")

print(result[["name", "salary"]])
````

## Method Chaining

Since most pandas methods return a Series or DataFrame, you can chain them one after another in a single line. This is called **method chaining**, and once you get comfortable with the individual pieces, it becomes a very natural way to write pandas code.

A simple example is `df.isna().sum()`. `isna()` checks for missing values and turns the whole DataFrame into `True`/`False` values. Chaining `.sum()` straight after counts up the `True`s in each column, giving you the number of missing values per column in one line.

````python
df.isna().sum()
````
