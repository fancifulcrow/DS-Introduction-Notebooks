## Creating Numpy Arrays

The array object in NumPy is called `ndarray`.

### Creating from a Python List and Tuple

You can create an Numpy `ndarray` from a regular Python list or tuple using the `array()` function. 

````python
arr = np.array([1, 2, 3, 4, 5])

print(arr)
print(type(arr))
````

## N-Dimensional Arrays

In computer programming, an array is a structure for storing and retrieving data. We often talk about an array as if it were a grid in space, with each cell storing one element of the data. For instance, if each element of the data were a number, we might visualize a “one-dimensional” array like a list:

|  1  |  5  |  2  |  0  |
|:---:|:---:|:---:|:---:|

A two-dimensional array would be like a table:

|  1  |  5  |  2  |  0  |
|:---:|:---:|:---:|:---:|
|  8  |  3  |  6  |  1  |
|  1  |  7  |  2  |  9  |

A three-dimensional array would be like a set of tables, perhaps stacked as though they were printed on separate pages.

NumPy `ndarray` have some restrictions. For instance:

- All elements of the array must be of the same type of data.
- Once created, the total size of the array can't change.
- The shape must be “rectangular”, not “jagged”; e.g., each row of a two-dimensional array must have the same number of columns.

## Data Types and Casting

NumPy supports many data types such as `int32`, `int64`, `float32`, `float64`, `bool`, and more. You can specify the data type when creating an array, or convert (cast) an existing array to a different type using `.astype()`.

This is useful when you need to control memory usage or prepare data for specific operations.

````python
arr = np.array([1.7, 2.5, 3.9, 4.1, 5.0])
print(arr.dtype)

# Cast from float to integer
arr = arr.astype(np.int64) # decimals are truncated, NOT rounded
print(arr.dtype)
print(arr)
````

````python
# Cast to boolean
arr = np.array([1, 0, 3, 0, 5])
print(arr.astype(bool))
````

## Indexing and Slicing

You can index and slice NumPy arrays in the same ways you can slice Python lists.

````python
# Indexing
arr = np.array([2, 4, 6, 8])

print(arr[2])
````

````python
# Slicing
arr = np.array([1, 2, 3, 4, 5, 6, 7])

print(arr[2:5])
````

````python
# Slicing
arr = np.array([1, 2, 3, 4, 5, 6, 7])

print(arr[2:])
````

````python
# Slicing
arr = np.array([1, 2, 3, 4, 5, 6, 7])

print(arr[:5])
````

````python
# Slicing
arr = np.array([1, 2, 3, 4, 5, 6, 7])

print(arr[2:-2])  # [3 4 5]
````

### Indexing and Slicing in Multidimensional Arrays

When working with multidimensional arrays, there are two common ways to access elements.

You can use nested indexing, which is similar to the one basic python uses.

````python
# This is a two dimensional array
arr = np.array([
    [0, 1, 2, 3], 
    [1, 4, 9, 0],
    [2, 5, 3, 5],
])

print(arr[1][2])
````

However we would much rather use:

````python
# This is a two dimensional array
arr = np.array([
    [0, 1, 2, 3], 
    [1, 4, 9, 0],
    [2, 5, 3, 5],
])

print(arr[1, 2])
````

This way is cleaner and more readable and is computationally faster.

Similarly, we can slice in this way:

````python
# This is a two dimensional array
arr = np.array([
    [0, 1, 2, 3], 
    [1, 4, 9, 0],
    [2, 5, 3, 5],
])

print(arr[1:, 1:3])
````

## Boolean Indexing

One of NumPy's most powerful features is **boolean indexing**. This is filtering an array by a condition. Instead of specifying index positions, you pass a condition and NumPy returns only the elements where the condition is `True`.

````python
arr = np.array([1, 2, 3, 4, 5, 6, 7])

print(arr > 3) # The condition produces a boolean array

print(arr[arr > 3]) # Use that condition to filter the array
````

````python
# You can combine conditions with & (and) and | (or)
arr = np.array([1, 2, 3, 4, 5, 6, 7])

# Elements greater than 1 AND less than 7 AND even
print(arr[(arr > 1) & (arr < 7) & (arr % 2 == 0)])
````

### `np.where()`

`np.where(condition, value_if_true, value_if_false)` lets you replace values in an array based on a condition. Think of it as a vectorized if-else.

````python
scores = np.array([30, 45, 55, 70, 85, 90])

# Label scores below 60 as 'low', 60 and above as 'high'
labels = np.where(scores < 60, 'low', 'high')
print(labels)
````

## Flattening and Reshaping

### Flattening

Both `.flatten()` and `.ravel()` methods are used to convert a multidimensional NumPy array into a 1D array.

The key difference is that `.flatten()` returns a **copy** of the array while `.ravel()` returns a **view** where possible.

````python
arr = np.array([
    [[0, 1, 2, 3], 
     [0, 4, 4, 0],
     [3, 3, 3, 3]],

    [[0, 1, 2, 3], 
     [0, 4, 9, 0],
     [3, 3, 3, 3]]
])

flat = arr.flatten()
print(flat)
````

````python
arr = np.array([
    [[0, 1, 2, 3], 
     [0, 4, 4, 0],
     [3, 3, 3, 3]],

    [[0, 1, 2, 3], 
     [0, 4, 9, 0],
     [3, 3, 3, 3]]
])

ravelled = arr.ravel()
print(ravelled)
````

### Reshaping

You can also reshape an array with the `.reshape()` method. Reshaping allows us to change the shape of a NumPy array without changing its data. However the size (i.e total number of elements) **must** remain the same for this to work.

````python
arr = np.array([1, 2, 3, 4, 5, 6])

arr = arr.reshape(2, 3)
print(arr)
````

## Copying Arrays

This is one of the most common sources of bugs for NumPy beginners. When you slice a NumPy array, you get a **view**, not a copy. This means changes to the slice will change the original array too.

Use `.copy()` when you want an independent copy.

````python
original = np.array([1, 2, 3, 4, 5])

view = original[1:3]
view[0] = 9

print("Original after modifying view:", original)
````

````python
original = np.array([1, 2, 3, 4, 5])

copy = original[1:3].copy()
copy[0] = 999

print("Original after modifying copy:", original)
````

## Sorting

`np.sort()` returns a sorted copy of an array. `np.argsort()` returns the **indices** that would sort the array and is useful when you need to know the rank or order of elements.

````python
arr = np.array([3, 1, 4, 5, 2, 7])

sorted_arr = np.sort(arr)

print(sorted_arr)
````

````python
# argsort is useful for sorting one array by the order of another
scores = np.array([85, 60, 90, 95, 70, 100])
names  = np.array(['Alice', 'Bob', 'Charlie', 'Diana', 'Eve', 'Frank'])

order = np.argsort(scores)
print("Indices that sort the array:", order)
print("Names ranked by score:", names[order])
````

## Stacking and Splitting

You can combine multiple arrays into one, or split one array into several pieces.

````python
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

# hstack: stack side by side (horizontally)
print("hstack:\n", np.hstack([a, b]))

# vstack: stack on top of each other (vertically)
print("vstack:\n", np.vstack([a, b]))

# concatenate: general-purpose joining along a specified axis
print("concatenate:\n", np.concatenate([a, b]))
````

````python
# split: split an array into equal parts
arr = np.array([1, 2, 3, 4, 5, 6])

parts = np.split(arr, 3)
print(parts)
````

## Random Numbers

The `np.random` module lets you generate random numbers and arrays.

Use `np.random.seed()` to make your random results **reproducible**. The same seed always produces the same random numbers.

````python
np.random.seed(42)  # set seed for reproducibility

# Random floats between 0 and 1
print("rand: ", np.random.rand(5))

# Random integers between 0 (inclusive) and 10 (exclusive)
print("randint: ", np.random.randint(0, 10, size=5))

# Random numbers from a standard normal distribution (mean=0, std=1)
print("randn: ", np.random.randn(5))
````

## Broadcasting

You can perform basic arithmetic operations (addition, subtraction, multiplication and division) with arrays of the same size. Calculations are done elementwise.

````python
a = np.array([0, 1, 2, 3, 4])
b = np.array([9, 8, 7, 6, 5])

print(a + b)
````

There are times when you might want to carry out an operation between an array and a single number (also called an operation between a vector and a scalar). You can perform this operation with:

````python
arr = np.array([1.0, 2.0])

arr = arr * 1.6

print(arr)
````

````python
arr = np.ones((2, 3))

arr += 4

print(arr)
````

NumPy understands that the operation should happen with each cell. That concept is called broadcasting. The dimensions of your array must be compatible. Two dimensions are compatible when they are equal or when one of them is 1.

Arrays do not need to have the same number of dimensions. Missing dimensions are treated as size 1, and the resulting array takes the shape formed by the largest size in each dimension.

````python
a = np.array([[1, 2, 3], [4, 5, 6]])  # shape: (2, 3)

b = np.array([10, 20, 30])  # shape: (3,). Treated as (1, 3) during broadcasting

print(a * b)  # shape of result: (2, 3)
````

## Maths

## Axis-Based Operations

Imagine a 2D array. You can apply functions along a specific axis. If `axis=0`, NumPy applies the operation down each column, reducing rows. If `axis=1`, it applies the operation across each row, reducing columns.

````python
arr = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

print("Array:\n", arr)
print("Sum of each column (axis=0):", np.sum(arr, axis=0))
print("Sum of each row (axis=1):", np.sum(arr, axis=1))
````

## Mini Project: Analysing Student Scores

Let's put everything together with a small example. We have test scores for 5 students across 3 subjects and want to compute useful statistics.

````python
# Rows = students, Columns = [Maths, English, Science]
scores = np.array([
    [78, 85, 90], # Alice
    [55, 60, 70], # Bob
    [90, 95, 88], # Charlie
    [40, 50, 45], # Diana
    [88, 76, 95], # Eve
])

names = np.array(['Alice', 'Bob', 'Charlie', 'Diana', 'Eve'])

# Average score per student (across columns)
avg_per_student = np.mean(scores, axis=1)

# Average score per subject (across rows)
avg_per_subject = np.mean(scores, axis=0)

# Who passed? (average >= 60)
passed = names[avg_per_student >= 60]
failed = names[avg_per_student < 60]

print("Scores:\n", scores)
print("\nAverage per student:", avg_per_student.round(2))
print("Average per subject:", avg_per_subject.round(2))
print("Highest score overall:", np.max(scores))
print("Students who passed (avg >= 60):", passed)
print("Students who failed:", failed)
````
